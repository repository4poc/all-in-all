```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

print(os.getenv("BASE_URL"))

client = OpenAI(
    base_url=os.getenv("BASE_URL"),
    api_key=os.getenv("API_KEY")
)

question="who is ...."

messages = [
    {
        "role": "user",
        "content": question
    }
]

response = client.chat.completions.create(
    model=os.getenv("MODEL"),
)
```
